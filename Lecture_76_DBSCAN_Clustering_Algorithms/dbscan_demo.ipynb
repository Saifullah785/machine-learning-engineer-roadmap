```python
# Import necessary libraries for DBSCAN, numpy, and plotting
from sklearn.cluster import DBSCAN
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Define the data points
X = np.array([[1, 2], [2, 2], [2, 3],[8, 7], [8, 8], [25, 80]])
```

```python
# Display the data points
X
```

```python
# Plot the data points
plt.scatter(X[:,0], X[:,1])
plt.show()
```

```python
# Apply DBSCAN to the data
db = DBSCAN(eps=10, min_samples=2)

db.fit(X)
```

```python
# Display the cluster labels assigned by DBSCAN
db.labels_
```

```python
# Import necessary libraries for plotting, DBSCAN, and creating data
import matplotlib.pyplot as plt
from sklearn.datasets import make_circles
from sklearn.cluster import DBSCAN
import numpy as np
```

```python
# Create a concentric circle dataset
X, _ = make_circles(n_samples=500, factor=.5, noise=.03, random_state=4)
```

```python
# Apply DBSCAN to the concentric circles dataset
dbscan = DBSCAN(eps=0.1, min_samples=5)
clusters = dbscan.fit_predict(X)
```

```python
# Plot the clustered concentric circles
plt.scatter(X[:, 0], X[:, 1], c=clusters, cmap='rainbow', marker='o')
plt.xlabel('Feature 0')
plt.ylabel('Feature 1')
plt.title('DBSCAN Clustering of concentric circles')
plt.show()
```
