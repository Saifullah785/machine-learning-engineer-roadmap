# **What is One Hot Encoding?**


---


One Hot Encoding is a method used to convert categorical data into a numerical format so that ML algorithms can process it.

Each category becomes a new binary column (0 or 1). Only one column is 1 per row, the rest are 0.

# **Why use it?**

ML models can’t process text labels directly.

Avoids false ordering (unlike Label Encoding).

Best for non-ordinal categorical features (e.g., color, city, gender).

# **3.One Hot Encoding using Sklearn**

```python
# Display the first few rows of the dataframe
df.head()
```

```python
# Split the data into training and testing sets
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(df.iloc[:,0:4],df.iloc[:,-1],test_size=0.2,random_state=2)
```

```python
# Display the first few rows of the training features
x_train.head()
```

```python
# Import OneHotEncoder from sklearn
from sklearn.preprocessing import OneHotEncoder
```

```python
# Initialize OneHotEncoder with drop='first' and dtype=np.int32
ohe = OneHotEncoder(drop='first',dtype=np.int32)
```

```python
# Apply one-hot encoding to 'fuel' and 'owner' columns in the training set
x_train_new = ohe.fit_transform(x_train[['fuel','owner']]).toarray()
```

```python
# Apply one-hot encoding to 'fuel' and 'owner' columns in the test set
x_test_new = ohe.fit_transform(x_test[['fuel','owner']]).toarray()
```

```python
# Check the shape of the one-hot encoded training data
x_train_new.shape
```

```python
# Horizontally stack the original numerical columns with the one-hot encoded columns for the training set
np.hstack((x_train[['brand','km_driven']].values,x_train_new))
```

# **4. One Hot Encoding with top Categories**

```python
# Calculate the value counts for the 'brand' column
counts = df['brand'].value_counts()
```

```python
# Get the number of unique brands and set a threshold for uncommon categories
df['brand'].nunique()
threshold = 100
```

```python
# Identify brands with counts less than or equal to the threshold
rep1 = counts[counts <= threshold].index
```

```python
# Perform one-hot encoding on the 'brand' column, grouping uncommon brands
pd.get_dummies(df['brand'].replace(rep1, 'uncommon')).sample(5)
```
