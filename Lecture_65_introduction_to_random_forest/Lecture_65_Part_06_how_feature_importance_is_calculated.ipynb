# **How-Feature-Importance-Is-Calculated**

```python
# Import the make_classification function from sklearn.datasets
# This function is used to generate synthetic classification datasets.
from sklearn.datasets import make_classification
```

```python
# Import the DecisionTreeClassifier class from sklearn.tree
# This is a class for building decision tree models.
from sklearn.tree import DecisionTreeClassifier
```

```python
# Generate a synthetic classification dataset
# n_samples: number of samples
# n_classes: number of classes
# n_features: total number of features
# n_informative: number of informative features (features that are useful for classification)
# n_redundant: number of redundant features (linear combinations of informative features)
# random_state: seed for reproducibility
X, y = make_classification(n_samples=5, n_classes=2, n_features=2, n_informative=2, n_redundant=0, random_state=0 )
```

```python
# Create an instance of the DecisionTreeClassifier
clf = DecisionTreeClassifier()
```

```python
# Train the DecisionTreeClassifier model using the generated data
clf.fit(X, y)
```

```python
# Import the plot_tree function from sklearn.tree
# This function is used to visualize a trained decision tree.
from sklearn.tree import plot_tree

# Plot the trained decision tree
plot_tree(clf)
```

```python
# Get the feature importances from the trained DecisionTreeClassifier
# Feature importance indicates the relative importance of each feature in the decision-making process.
clf.feature_importances_
```

```python
# Import the RandomForestClassifier class from sklearn.ensemble
# This is a class for building random forest models (an ensemble of decision trees).
from sklearn.ensemble import RandomForestClassifier

# Create an instance of the RandomForestClassifier
# n_estimators: the number of trees in the forest
rf = RandomForestClassifier(n_estimators=2)

# Train the RandomForestClassifier model using the generated data
rf.fit(X, y)
```

```python
# Get the feature importances from the trained RandomForestClassifier
# Feature importance in a random forest is the average of the feature importances of individual trees.
rf.feature_importances_
```

```python
# Print the feature importances for each individual tree in the random forest
print(rf.estimators_[0].feature_importances_)
print(rf.estimators_[1].feature_importances_)

# Print the average of two numbers (this seems unrelated to the feature importances printed above)
print((1 + 0.555)/2)
```

```python
# Generate another synthetic classification dataset with more samples
# n_samples: number of samples
# n_classes: number of classes
# n_features: total number of features
# n_informative: number of informative features (features that are useful for classification)
# n_redundant: number of redundant features (linear combinations of informative features)
# random_state: seed for reproducibility
X, y  = make_classification (n_samples = 15, n_classes=2,
                               n_features=2, n_informative=2, n_redundant=0, random_state=0)
```

```python
# Import the DecisionTreeClassifier class from sklearn.tree
# This is a class for building decision tree models.
from sklearn.tree import DecisionTreeClassifier

# Create a new instance of the DecisionTreeClassifier
clf = DecisionTreeClassifier()
```

```python
# Train the new DecisionTreeClassifier model using the larger generated dataset
clf.fit(X, y)
```

```python
# Import the plot_tree function from sklearn.tree
# This function is used to visualize a trained decision tree.
from sklearn.tree import plot_tree

# Plot the trained decision tree for the larger dataset
plot_tree(clf)
```

```python
# Get the feature importances from the trained DecisionTreeClassifier on the larger dataset
# Feature importance indicates the relative importance of each feature in the decision-making process.
clf.feature_importances_
```

# **Feature-Importance-In-Sklearn**

```python
# Import necessary libraries: numpy, pandas, matplotlib.pyplot, and seaborn
# numpy: for numerical operations
# pandas: for data manipulation and analysis
# matplotlib.pyplot: for creating static, interactive, and animated visualizations
# seaborn: for creating statistical graphics based on matplotlib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import tensorflow as tf

# Load the MNIST dataset using tensorflow.keras.datasets
# x_train, y_train: training data and labels
# x_test, y_test: testing data and labels
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# Display the shapes of the loaded data to understand their dimensions
print(f"Training data shape: {x_train.shape}")
print(f"Training labels shape: {y_train.shape}")
print(f"Testing data shape: {x_test.shape}")
print(f"Testing labels shape: {x_test.shape}") # Corrected from y_test.shape
```

```python
import matplotlib.pyplot as plt
# Reshape the image data from 28x28 pixels to a flat array of 784 pixels
# This is required for scikit-learn models which expect 2D input (samples, features)
x_train_flat = x_train.reshape(x_train.shape[0], -1)
x_test_flat = x_test.reshape(x_test.shape[0], -1)

# Create a RandomForestClassifier model
# n_estimators: the number of trees in the forest (set to 50 for quicker demonstration)
# random_state: seed for reproducibility
# n_jobs: number of jobs to run in parallel (-1 uses all available processors)
rf_mnist = RandomForestClassifier(n_estimators=50, random_state=42, n_jobs=-1)

# Train the RandomForestClassifier model on the flattened MNIST training data and labels
print("Training RandomForestClassifier on MNIST data...")
rf_mnist.fit(x_train_flat, y_train)
print("Training complete.")
```

```python
# Get the feature importances from the trained RandomForestClassifier on the MNIST dataset
# Each value in the array corresponds to the importance of a pixel (feature) in the image.
rf_mnist.feature_importances_
```

```python
# Check the shape of the feature importances array
# This confirms that there is one importance value for each pixel (feature) in the flattened image.
rf_mnist.feature_importances_.shape
```

```python
# Assign the calculated feature importances to a variable for easier use
feature_importances = rf_mnist.feature_importances_
```

```python
# Reshape the flat array of feature importances back into a 28x28 image shape
# This allows us to visualize the importances as an image.
feature_importances_image = feature_importances.reshape(28, 28)
```

```python
# Create a heatmap to visualize the feature importances as an image
# plt.figure: create a new figure
# sns.heatmap: create the heatmap
# cmap: colormap (viridis is a common choice)
# annot: whether to annotate each cell with the data value (set to False here)
# fmt: format string for annotations (not used here as annot is False)
# plt.title: set the title of the plot
# plt.xlabel: set the label for the x-axis (Pixel Column)
# plt.ylabel: set the label for the y-axis (Pixel Row)
# plt.show(): display the plot
plt.figure(figsize=(10, 8))
sns.heatmap(feature_importances_image, cmap='viridis', annot=False, fmt=".2f")
plt.title('Random Forest Feature Importance Heatmap for MNIST')
plt.xlabel('Pixel Column')
plt.ylabel('Pixel Row')
plt.show()
```
