# importing the libraries

```python
# Import necessary libraries
import numpy as np
import pandas as pd
```

# Loading & Preprocess the Data

```python
# Load the dataset from a URL
url = 'https://raw.githubusercontent.com/Saifullah785/machine-learning-engineer-roadmap/bdef780aee6627f05086800c969ac8507634a5ff/Lecture_13_end_to_end_toy_project/placement.csv'
df = pd.read_csv(url)
```

```python
df.head()
```

```python
# Remove the first column
df = df.iloc[:,1:]
```

```python
# Display the first few rows of the dataframe
df.head()
```

```python
# Display the shape of the dataframe
df.shape
```

```python
# Display information about the dataframe
df.info()
```

# Exploratory Data Analysis (EDA)

```python
# Import matplotlib for plotting
import matplotlib.pyplot as plt
```

```python
# Create a scatterplot of cgpa vs iq, colored by placement
plt.scatter(df['cgpa'],df['iq'],c=df['placement'])
```

# EXtract input and Output Cols

```python
# Extract input features (x) and output variable (y)
x = df.iloc[:,0:2]
y= df.iloc[:,-1]
```

```python
# Display the input features
x
```

```python
# Display the shape of the output variable
y.shape
```

# Train Test Split

```python
# Import train_test_split for splitting data
from sklearn.model_selection import train_test_split
```

```python
# Split the data into training and testing sets
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.1)
```

```python
# Display the training data for input features
x_train
```

```python
# Display the testing data for input features
x_test
```

# Scale the Values

```python
# Import StandardScaler for feature scaling
from sklearn.preprocessing import StandardScaler
```

```python
# Create a StandardScaler object
scaler = StandardScaler()
```

```python
# Fit and transform the training data
x_train = scaler.fit_transform(x_train)
```

```python
# Display the scaled training data
x_train
```

```python
# Fit and transform the testing data
x_test = scaler.fit_transform(x_test)
```

```python
# Display the scaled testing data
x_test
```

# Train the model

```python
# Import LogisticRegression model
from sklearn.linear_model import LogisticRegression
```

```python
# Create a LogisticRegression model object
clf = LogisticRegression()
```

```python
# Train the LogisticRegression model
clf.fit(x_train,y_train)
```

# Evaluating the model

```python
# Make predictions on the test data
y_pred =clf.predict(x_test)
```

```python
# Display the actual test data output
y_test
```

# Accuracy Score

```python
# Import accuracy_score for evaluating the model
from sklearn.metrics import accuracy_score
```

```python
# Calculate and display the accuracy score
accuracy_score(y_test,y_pred)*100
```

```python
# Import plot_decision_regions for visualizing decision boundaries
from mlxtend.plotting import plot_decision_regions
```

```python
# Plot the decision regions of the trained model
plot_decision_regions(x_train, y_train.values, clf=clf,legend=2)
```
