# ⚠️ CHALLENGE #1: Poor Quality Data – The Silent Killer of Models ☠️

**The Problem:** Your model is only as good as the data you feed it. Real-world data is rarely perfect. It's often riddled with missing values, noise, and outliers – each a potential landmine for your ML pipeline.

*   **Missing Values:** Gaps in your data. Should you fill them? With what? Ignoring them can lead to biased models or outright errors.
*   **Noise:** Random errors or irrelevant information that obscure the true patterns in your data.
*   **Outliers:** Data points that significantly deviate from the majority. Are they errors, or rare, but important events?

**The Consequence:** Models trained on poor quality data learn poor quality patterns. This leads to inaccurate predictions, unreliable insights, and ultimately, a failed ML project.

### 💻 **Practical Deep Dive: Unmasking Data Imperfections**

Let's load a classic dataset – the **Pima Indians Diabetes Database**. While seemingly clean, it has a notorious hidden secret: missing values encoded as zeros! We'll expose these imperfections and learn how to handle them.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set plotting style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = [10, 6]
plt.rcParams['font.size'] = 12

# Load the dataset
# The dataset is available through UCI Machine Learning Repository or directly via a common URL
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.csv"
column_names = ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age', 'Outcome']
df = pd.read_csv(url, names=column_names)

print("Dataset loaded successfully!")
print("First 5 rows of the dataset:")
display(df.head())

print("\nDataset Information:")
df.info()

print("\nDescriptive Statistics:")
display(df.describe())
```

#### **The Hidden Secret: Zeros as Missing Values!** 🤫

Notice the `min` values in the `describe()` output for columns like `Glucose`, `BloodPressure`, `SkinThickness`, `Insulin`, and `BMI`. They are `0`! Biologically, these values (except for `Pregnancies`) cannot be zero. A `Glucose` of 0 means death, a `BMI` of 0 is impossible. These zeros are placeholders for missing data.

This is a common issue in real-world datasets: missing values aren't always `NaN` or empty strings; they can be encoded as 0s, 999s, or other arbitrary values. Our first task is to correctly identify and treat these.

```python
# Columns where 0s represent missing values
columns_with_zero_as_missing = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

# Replace 0s with NaN in these columns
df[columns_with_zero_as_missing] = df[columns_with_zero_as_missing].replace(0, np.nan)

print("Updated Descriptive Statistics after replacing 0s with NaN:")
display(df.describe())

print("\nNumber of missing values per column after conversion:")
display(df.isnull().sum())
```

#### **Visualizing Missing Data: A Heatmap of Emptiness** 📊

A visual representation often makes data problems stand out. A heatmap of missing values can quickly show patterns and the extent of data loss.

```python
plt.figure(figsize=(12, 7))
sns.heatmap(df.isnull(), cbar=False, cmap='viridis')
plt.title('Heatmap of Missing Values (NaN)')
plt.xlabel('Features')
plt.ylabel('Samples')
plt.show()
```

#### **Handling Missing Values: Imputation Strategies** 🩹

Now that we've identified the missing values, what do we do? We can't just delete all rows with NaNs, as we'd lose too much valuable data. Imputation is the process of filling in these gaps.

Common strategies include:
*   **Mean/Median Imputation:** Replace NaNs with the mean or median of the column. (Good for numerical data, simple, but reduces variance).
*   **Mode Imputation:** Replace NaNs with the most frequent value. (Good for categorical data).
*   **Forward/Backward Fill:** Propagate the last/next valid observation. (Good for time-series data).
*   **K-Nearest Neighbors (KNN) Imputation:** Impute based on values from similar data points.
*   **Model-Based Imputation:** Use another ML model to predict missing values.

For our numerical features, **median imputation** is often preferred over mean imputation because it is less sensitive to outliers. Let's apply it.

```python
# Impute missing values with the median of their respective columns
for column in columns_with_zero_as_missing:
    median_val = df[column].median()
    df[column].fillna(median_val, inplace=True)

print("Number of missing values after median imputation:")
display(df.isnull().sum())

print("\nDescriptive Statistics after imputation:")
display(df.describe())
```

#### **Dealing with Outliers: Taming the Wild Data Points** 🌲

Outliers are data points that lie an abnormal distance from other values. They can skew our model's understanding of the data, especially for algorithms sensitive to extreme values (like linear regression or KMeans).

**Identification Methods:**
*   **Z-score:** For normally distributed data.
*   **IQR (Interquartile Range):** Robust for skewed data.
*   **Visualizations:** Box plots, scatter plots.
*   **Isolation Forest, One-Class SVM:** Advanced ML-based methods.

**Handling Strategies:**
*   **Removal:** Delete the outlier data points (only if they are errors or rare enough not to impact the distribution too much).
*   **Capping/Winsorization:** Replace outliers with a maximum or minimum value (e.g., 99th percentile).
*   **Transformation:** Apply logarithmic or other transformations to reduce the impact of skewed distributions.

Let's visualize some distributions with box plots to identify potential outliers and then use IQR to cap them for a more robust dataset.

```python
# Visualize distributions and outliers with box plots
fig, axes = plt.subplots(nrows=3, ncols=3, figsize=(18, 15))
axes = axes.flatten()

for i, column in enumerate(df.columns[:-1]): # Exclude 'Outcome' for outlier analysis
    sns.boxplot(y=df[column], ax=axes[i], color='skyblue')
    axes[i].set_title(f'Box Plot of {column}')
    axes[i].set_ylabel('')

# Remove empty subplots if any
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.suptitle('Feature Distributions with Outliers Before Capping', y=1.02, fontsize=16)
plt.show()
```

```python
# Capping outliers using the IQR method
# Define a function to cap outliers
def cap_outliers_iqr(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return series.clip(lower=lower_bound, upper=upper_bound)

# Apply outlier capping to numerical columns (excluding 'Outcome')
for column in df.columns[:-1]:
    df[column] = cap_outliers_iqr(df[column])

print("Descriptive Statistics after outlier capping:")
display(df.describe())

# Visualize distributions after capping
fig, axes = plt.subplots(nrows=3, ncols=3, figsize=(18, 15))
axes = axes.flatten()

for i, column in enumerate(df.columns[:-1]):
    sns.boxplot(y=df[column], ax=axes[i], color='lightcoral')
    axes[i].set_title(f'Box Plot of {column} After Capping')
    axes[i].set_ylabel('')

# Remove empty subplots if any
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.suptitle('Feature Distributions After Outlier Capping', y=1.02, fontsize=16)
plt.show()

print("\n🔥 **Key takeaway for Challenge #1:** Data cleaning is not glamorous, but it's the bedrock of any successful ML project. Ignoring these steps is like building a skyscraper on quicksand!")
```

# ⚖️ CHALLENGE #2: Imbalanced Data – The Accuracy Illusion 🎭

**The Problem:** Imagine you're building a model to detect a rare disease. If only 1% of your dataset has the disease, a model that *always* predicts 'no disease' will achieve 99% accuracy! While statistically impressive, this model is utterly useless. This is the danger of imbalanced data: accuracy becomes a misleading metric, and your model fails to correctly identify the minority class, which is often the one you care about most.

**The Consequence:** Models trained on imbalanced datasets become biased towards the majority class. They perform poorly on the minority class, leading to missed opportunities, poor decision-making, and potentially harmful outcomes (e.g., failing to detect fraud, disease, or critical system failures).

### 💻 **Practical Deep Dive: Unmasking and Balancing Class Skew**

Let's check the balance of our `Outcome` variable (diabetes vs. no diabetes) and see how a simple model performs on it without any balancing strategies.

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
from sklearn.linear_model import LogisticRegression
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import plotly.express as px # Added import statement for plotly.express

# Prepare data for modeling
X = df.drop('Outcome', axis=1)
y = df['Outcome']

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

print("Class distribution in original dataset:")
display(y.value_counts(normalize=True))

# Visualize class distribution
fig = px.pie(names=y.value_counts().index, values=y.value_counts().values, title='Original Class Distribution of Outcome')
fig.update_traces(textposition='inside', textinfo='percent+label')
fig.show()

print("\n--- Training a simple Logistic Regression model WITHOUT balancing ---")
# Train a simple model without handling imbalance
model_unbalanced = LogisticRegression(solver='liblinear', random_state=42)
model_unbalanced.fit(X_train, y_train)

# Make predictions
y_pred_unbalanced = model_unbalanced.predict(X_test)

# Evaluate the model
print(f"Accuracy (unbalanced): {accuracy_score(y_test, y_pred_unbalanced):.4f}")
print(f"Precision (unbalanced): {precision_score(y_test, y_pred_unbalanced):.4f}")
print(f"Recall (unbalanced): {recall_score(y_test, y_pred_unbalanced):.4f}")
print(f"F1-Score (unbalanced): {f1_score(y_test, y_pred_unbalanced):.4f}")

cm_unbalanced = confusion_matrix(y_test, y_pred_unbalanced)
print("\nConfusion Matrix (unbalanced):\n", cm_unbalanced)

# Plot Confusion Matrix using Seaborn
plt.figure(figsize=(6, 5))
sns.heatmap(cm_unbalanced, annot=True, fmt='d', cmap='Blues', xticklabels=['No Diabetes', 'Diabetes'], yticklabels=['No Diabetes', 'Diabetes'])
plt.title('Confusion Matrix (Unbalanced Data)')
plt.ylabel('Actual')
plt.xlabel('Predicted')
plt.show()
```

#### **Interpreting the Results: The Deceptive Accuracy**

Notice how the accuracy might look decent, but the **recall** for the minority class (Outcome=1, Diabetes) is likely low. This means our model is failing to identify a significant portion of actual positive cases, which can be critical in many applications like disease detection.

### **Handling Imbalance: Resampling Strategies** 🛠️

To address class imbalance, we can use resampling techniques:

*   **Oversampling:** Increasing the number of instances in the minority class (e.g., using SMOTE - Synthetic Minority Over-sampling Technique).
*   **Undersampling:** Decreasing the number of instances in the majority class.

For this lecture, we'll use **SMOTE** to synthesize new samples for the minority class, helping our model learn better patterns for 'diabetes' cases without losing information from the majority class.

```python
from imblearn.over_sampling import SMOTE
from collections import Counter

# Before SMOTE
print(f"Original training set shape: {X_train.shape}, Labels: {y_train.shape}")
print(f"Original class distribution: {Counter(y_train)}")

# Apply SMOTE to the training data
sm = SMOTE(random_state=42)
X_train_res, y_train_res = sm.fit_resample(X_train, y_train)

# After SMOTE
print(f"Resampled training set shape: {X_train_res.shape}, Labels: {y_train_res.shape}")
print(f"Resampled class distribution: {Counter(y_train_res)}")

# Visualize class distribution after SMOTE using Plotly
labels = ['No Diabetes (0)', 'Diabetes (1)']

fig = make_subplots(rows=1, cols=2, specs=[[{'type':'domain'}, {'type':'domain'}]],
                    subplot_titles=['Original Training Data', 'SMOTE Resampled Training Data'])

fig.add_trace(go.Pie(labels=labels, values=[Counter(y_train)[0], Counter(y_train)[1]], name="Original"), 1, 1)
fig.add_trace(go.Pie(labels=labels, values=[Counter(y_train_res)[0], Counter(y_train_res)[1]], name="SMOTE"), 1, 2)

fig.update_layout(title_text='Class Distribution Before and After SMOTE', title_x=0.5)
fig.show()

print("\n--- Training Logistic Regression model WITH SMOTE balancing ---")
# Train a model with the balanced data
model_balanced = LogisticRegression(solver='liblinear', random_state=42)
model_balanced.fit(X_train_res, y_train_res)

# Make predictions
y_pred_balanced = model_balanced.predict(X_test)

# Evaluate the model
print(f"Accuracy (balanced): {accuracy_score(y_test, y_pred_balanced):.4f}")
print(f"Precision (balanced): {precision_score(y_test, y_pred_balanced):.4f}")
print(f"Recall (balanced): {recall_score(y_test, y_pred_balanced):.4f}")
print(f"F1-Score (balanced): {f1_score(y_test, y_pred_balanced):.4f}")

cm_balanced = confusion_matrix(y_test, y_pred_balanced)
print("\nConfusion Matrix (balanced):\n", cm_balanced)

# Plot Confusion Matrix using Seaborn after balancing
plt.figure(figsize=(6, 5))
sns.heatmap(cm_balanced, annot=True, fmt='d', cmap='Greens', xticklabels=['No Diabetes', 'Diabetes'], yticklabels=['No Diabetes', 'Diabetes'])
plt.title('Confusion Matrix (Balanced Data with SMOTE)')
plt.ylabel('Actual')
plt.xlabel('Predicted')
plt.show()

print("\n🔥 **Key takeaway for Challenge #2:** Always check your class distribution! High accuracy on imbalanced data can be a mirage. Metrics like Precision, Recall, and F1-Score provide a truer picture of your model's performance, especially for the critical minority class.")
```

# 🧠 CHALLENGE #3: Overfitting & Underfitting – The Goldilocks Zone of ML Models 🐻

**The Problem:** Training an ML model is like teaching a student. An **underfit** model is like a student who didn't study enough – they don't grasp the basic concepts and perform poorly on both practice (training) and exams (test data). An **overfit** model is like a student who memorized every single answer to past exams but doesn't understand the underlying principles – they ace the practice tests but fail miserably on new, unseen exam questions.

*   **Underfitting:** Occurs when the model is too simple to capture the underlying patterns in the training data. High bias, low variance.
*   **Overfitting:** Occurs when the model is too complex and learns the noise and specific details of the training data, rather than the general trends. Low bias, high variance.

**The Consequence:** Both underfitting and overfitting lead to poor generalization. An underfit model won't solve your problem at all. An overfit model will give you a false sense of security during development, only to collapse in production when faced with real-world data it hasn't 'memorized'.

### 💻 **Practical Deep Dive: Finding the Sweet Spot**

Let's demonstrate overfitting and underfitting using polynomial regression and visualize learning curves to understand how model complexity affects performance on both training and test sets. We will use the cleaned and balanced dataset from the previous steps.

```python
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.model_selection import learning_curve
from sklearn.metrics import mean_squared_error

# Import imblearn's pipeline for SMOTE integration and SMOTE itself
from imblearn.pipeline import Pipeline as ImbPipeline
from imblearn.over_sampling import SMOTE

# For simplicity and clear visualization of overfitting/underfitting, let's use a subset of features
# and revert to the original (imputed but not capped) df for this example,
# as capping can sometimes obscure the effect on specific features for this demo.
# However, for a real project, you'd use the fully cleaned df.

# Let's use a simpler dataset for polynomial regression visualization example, maybe just Glucose and BMI against Outcome.
# But to maintain consistency with df, we'll use it but select features to make the visualization clearer

X_poly = df[['Glucose', 'BMI']]
y_poly = df['Outcome']

X_train_poly, X_test_poly, y_train_poly, y_test_poly = train_test_split(X_poly, y_poly, test_size=0.3, random_state=42)

# --- Underfitting Example: A very simple model (low degree polynomial) ---

print("\n--- Demonstrating Underfitting (Low Complexity Model) ---")
poly_pipeline_underfit = ImbPipeline([ # Changed to ImbPipeline
    ('scaler', StandardScaler()),
    ('poly', PolynomialFeatures(degree=1)), # Linear model (degree 1) for underfitting
    ('smote', SMOTE(random_state=42)),       # Add SMOTE here
    ('regressor', LogisticRegression(solver='liblinear', random_state=42))
])
poly_pipeline_underfit.fit(X_train_poly, y_train_poly)

y_pred_train_underfit = poly_pipeline_underfit.predict(X_train_poly)
y_pred_test_underfit = poly_pipeline_underfit.predict(X_test_poly)

print(f"Underfit Model - Train Accuracy: {accuracy_score(y_train_poly, y_pred_train_underfit):.4f}")
print(f"Underfit Model - Test Accuracy: {accuracy_score(y_test_poly, y_pred_test_underfit):.4f}")

# --- Overfitting Example: A very complex model (high degree polynomial) ---

print("\n--- Demonstrating Overfitting (High Complexity Model) ---")
poly_pipeline_overfit = ImbPipeline([ # Changed to ImbPipeline
    ('scaler', StandardScaler()),
    ('poly', PolynomialFeatures(degree=10)), # High degree polynomial for overfitting
    ('smote', SMOTE(random_state=42)),        # Add SMOTE here
    ('regressor', LogisticRegression(solver='liblinear', random_state=42, max_iter=1000)) # Increased max_iter for convergence
])
poly_pipeline_overfit.fit(X_train_poly, y_train_poly)

y_pred_train_overfit = poly_pipeline_overfit.predict(X_train_poly)
y_pred_test_overfit = poly_pipeline_overfit.predict(X_test_poly)

print(f"Overfit Model - Train Accuracy: {accuracy_score(y_train_poly, y_pred_train_overfit):.4f}")
print(f"Overfit Model - Test Accuracy: {accuracy_score(y_test_poly, y_pred_test_overfit):.4f}")

# --- Just Right Example: A balanced model (medium degree polynomial) ---

print("\n--- Demonstrating 'Just Right' (Balanced Complexity Model) ---")
poly_pipeline_justright = ImbPipeline([ # Changed to ImbPipeline
    ('scaler', StandardScaler()),
    ('poly', PolynomialFeatures(degree=2)), # Medium degree polynomial
    ('smote', SMOTE(random_state=42)),       # Add SMOTE here
    ('regressor', LogisticRegression(solver='liblinear', random_state=42))
])
poly_pipeline_justright.fit(X_train_poly, y_train_poly)

y_pred_train_justright = poly_pipeline_justright.predict(X_train_poly)
y_pred_test_justright = poly_pipeline_justright.predict(X_test_poly)

print(f"Just Right Model - Train Accuracy: {accuracy_score(y_train_poly, y_pred_train_justright):.4f}")
print(f"Just Right Model - Test Accuracy: {accuracy_score(y_test_poly, y_pred_test_justright):.4f}")
```

#### **Visualizing Learning Curves** 📈

Learning curves are powerful diagnostic tools that show how the model's performance (on both training and validation sets) changes with the amount of training data. They help us identify if the model suffers from high bias (underfitting) or high variance (overfitting).

```python
def plot_learning_curve(estimator, title, X, y, axes=None, ylim=None, cv=None,
                        n_jobs=None, train_sizes=np.linspace(.1, 1.0, 5)):
    """
    Generate 3 plots: the test and training learning curve, the training
    samples vs. fit times curve, and the fit times vs. score curve.

    Parameters
    ----------
    estimator : object type that implements the "fit" and "predict" methods
        An object of that type which is cloned for each validation.

    title : string
        Title for the chart.

    X : array-like, shape (n_samples, n_features)
        Training vector, where n_samples is the number of samples and
        n_features is the number of features.

    y : array-like, shape (n_samples) or (n_samples, n_features),
        Target vector as relationship to X.

    axes : array-like of shape (3,), optional
        Axes to plot the curves on.

    ylim : tuple, shape (ymin, ymax), optional
        Defines minimum and maximum yvalues plotted.

    cv : int, cross-validation generator or an iterable, optional
        Determines the cross-validation splitting strategy.
        Possible inputs for cv are:
          - None, to use the default 5-fold cross-validation,
          - integer, to specify the number of folds.
          - :term:`CV splitter`,
          - An iterable yielding (train, test) splits as arrays of indices.

        For integer/None inputs, if ``y`` is binary or multiclass, :class:`StratifiedKFold` is used.
        If the estimator is not a classifier or if ``y`` is neither binary nor multiclass,
        :class:`KFold` is used.

        Refer :ref:`User Guide <cross_validation>` for the various
        cross-validators that can be used.

    n_jobs : int or None, optional (default=None)
        Number of jobs to run in parallel. Training the estimator and computing the score are parallelized over the cross-validation splits.
        ``None`` means 1 unless in a :obj:`joblib.parallel_backend` context.
        ``-1`` means using all processors.

    train_sizes : array-like, shape (n_ticks,), optional
        Relative or absolute numbers of training examples that will be used to generate the learning curve.
        If the ``dtype`` is float, it is regarded as a fraction of the maximum training size (that is determined by the selected validation method), i.e. it has to be within (0, 1].
        Otherwise it is interpreted as absolute sizes of training samples.
        Note that for classification the number of samples usually have to be sufficient to contain at least one sample from each class.
        (default: np.linspace(0.1, 1.0, 5))
    """
    if axes is None:
        _, axes = plt.subplots(1, 3, figsize=(20, 5))

    axes[0].set_title(title)
    if ylim is not None:
        axes[0].set_ylim(*ylim)
    axes[0].set_xlabel("Training examples")
    axes[0].set_ylabel("Score (Accuracy)")

    train_sizes, train_scores, test_scores, fit_times, _ = \
        learning_curve(estimator, X, y, cv=cv, n_jobs=n_jobs,
                       train_sizes=train_sizes, return_times=True, scoring='accuracy')
    train_scores_mean = np.mean(train_scores, axis=1)
    train_scores_std = np.std(train_scores, axis=1)
    test_scores_mean = np.mean(test_scores, axis=1)
    test_scores_std = np.std(test_scores, axis=1)
    fit_times_mean = np.mean(fit_times, axis=1)
    fit_times_std = np.std(fit_times, axis=1)

    # Plot learning curve
    axes[0].grid()
    axes[0].fill_between(train_sizes, train_scores_mean - train_scores_std,
                         train_scores_mean + train_scores_std, alpha=0.1,
                         color="r")
    axes[0].fill_between(train_sizes, test_scores_mean - test_scores_std,
                         test_scores_mean + test_scores_std, alpha=0.1,
                         color="g")
    axes[0].plot(train_sizes, train_scores_mean, 'o-', color="r",
                 label="Training score")
    axes[0].plot(train_sizes, test_scores_mean, 'o-', color="g",
                 label="Cross-validation score")
    axes[0].legend(loc="best")

    # Plot n_samples vs fit_times
    axes[1].grid()
    axes[1].plot(train_sizes, fit_times_mean, 'o-')
    axes[1].fill_between(train_sizes, fit_times_mean - fit_times_std,
                         fit_times_mean + fit_times_std, alpha=0.1)
    axes[1].set_xlabel("Training examples")
    axes[1].set_ylabel("fit_times")
    axes[1].set_title("Scalability of the model")

    # Plot fit_time vs score
    axes[2].grid()
    axes[2].plot(fit_times_mean, test_scores_mean, 'o-')
    axes[2].fill_between(fit_times_mean, test_scores_mean - test_scores_std,
                         test_scores_mean + test_scores_std, alpha=0.1)
    axes[2].set_xlabel("fit_times")
    axes[2].set_ylabel("Score")
    axes[2].set_title("Performance of the model")

    return plt

# Prepare data again using the resampled data (y_train_res, y_test are actual labels for testing)
# Corrected: Pass the non-resampled training data to learning_curve, as SMOTE is now part of the pipelines.
X_train_lc = X_train_poly
y_train_lc = y_train_poly

# Corrected: Create a 3x3 grid of subplots to pass 3 axes to each learning curve call
fig, axes = plt.subplots(3, 3, figsize=(18, 15)) # Changed from 3,2 to 3,3

# Learning Curve for Underfit Model (degree=1)
plot_learning_curve(poly_pipeline_underfit, "Learning Curve (Underfit - Degree 1)",
                    X_train_lc, y_train_lc, axes=axes[0,:], ylim=(0.6, 1.01), cv=5, n_jobs=-1)

# Learning Curve for Just Right Model (degree=2)
plot_learning_curve(poly_pipeline_justright, "Learning Curve (Just Right - Degree 2)",
                    X_train_lc, y_train_lc, axes=axes[1,:], ylim=(0.6, 1.01), cv=5, n_jobs=-1)

# Learning Curve for Overfit Model (degree=10)
plot_learning_curve(poly_pipeline_overfit, "Learning Curve (Overfit - Degree 10)",
                    X_train_lc, y_train_lc, axes=axes[2,:], ylim=(0.6, 1.01), cv=5, n_jobs=-1)

plt.tight_layout()
plt.show()

print("\n🔥 **Key takeaway for Challenge #3:** The goal is to find a model complexity that captures patterns without memorizing noise. Learning curves are your best friend here: ")
print("   - **Underfitting (High Bias):** Both training and cross-validation scores are low and converge at a low value.")
print("   - **Overfitting (High Variance):** Training score is high, but cross-validation score is significantly lower. There's a large gap between the two.")
print("   - **Just Right:** Both scores are high and close to each other, with the gap between them being small.")
```

# 🔍 CHALLENGE #4: Feature Selection – Less Can Be More 🧘‍♀️

**The Problem:** Not all features are created equal. Including irrelevant, redundant, or noisy features in your model can be detrimental. It's like asking a detective to find a culprit in a room full of distractions – they might get overwhelmed, focus on the wrong clues, and miss the real perpetrator.

*   **Irrelevant Features:** Features that have no bearing on the target variable. They add noise and computational cost.
*   **Redundant Features:** Features that carry similar information as other features. They can inflate model complexity and cause multicollinearity issues.
*   **Curse of Dimensionality:** As the number of features (dimensions) increases, the data becomes sparser, making it harder for models to find meaningful patterns.

**The Consequence:** Too many features can lead to:
*   **Overfitting:** The model tries to learn from noise.
*   **Increased Training Time:** More data to process.
*   **Reduced Model Interpretability:** Harder to understand why a model makes certain predictions.
*   **Decreased Generalization:** Poor performance on unseen data.

# ⚡ CHALLENGE #5: Data Leakage – The Cheating Model Syndrome 🕵️‍♀️

**The Problem:** Imagine you're grading an exam, but you accidentally give the student the answer key beforehand. Unsurprisingly, they score perfectly! However, when they take a real, unseen exam, they fail. This is exactly what data leakage is: your model 'sees' information that it shouldn't have access to during training, especially information that would not be available at prediction time.

Data leakage happens when information from your test set (or future data) inadvertently 'leaks' into your training set.

**Common Scenarios for Data Leakage:**

*   **Target Leakage:** A feature that is directly or indirectly influenced by the target variable, and this influence wouldn't be present at prediction time. For example, in a loan default prediction model, if a feature like 'late payment fee collected' is used, it directly indicates a default, but wouldn't be known *before* the loan defaults.
*   **Train-Test Contamination:** Performing data preprocessing steps (like scaling, imputation, or feature selection) on the *entire* dataset before splitting into train and test sets. This allows information from the test set to influence the transformations learned from the training set.
*   **Time Series Leakage:** Using future information to predict past events (e.g., using tomorrow's stock price to predict yesterday's).

**The Consequence:** Data leakage leads to overly optimistic model performance during development and testing. Your model will appear to be highly accurate, but it will dramatically underperform in a real-world production environment where the leaked information is not available. This is a silent killer, as it's hard to detect and can lead to disastrous business decisions.

### 💡 **Key Principle: Process, Then Split!**

To avoid train-test contamination, always split your data into training and test sets *before* performing any data preprocessing steps that involve learning parameters from the data (like `StandardScaler.fit`, `SimpleImputer.fit`, or `SMOTE.fit_resample`). Apply these learned transformations *separately* to the training and test sets.

#### **The Golden Rule: Information Flow**

Think of it as a strict firewall: **Information from the future or from the test set must *never* flow into the training process or feature engineering.**

If you find your model performing 'too well' on validation data, leakage should be one of the first suspects! It's often subtle and requires careful scrutiny of your data pipeline.

# ⚙️ MINI PROJECT: Building a Resilient ML Pipeline 💪

Now, let's synthesize all the lessons learned into a single, robust machine learning pipeline. We'll take our raw (initial) data, apply the cleaning, imputation, balancing, and feature selection techniques, then train and evaluate a model. This integrated approach is crucial for building production-ready ML systems.

### **Pipeline Steps:**

1.  **Load Raw Data:** Start from scratch with the original, imperfect dataset.
2.  **Data Cleaning & Imputation:** Replace 0s with NaNs and impute missing values (median).
3.  **Outlier Capping:** Apply IQR-based capping to mitigate extreme values.
4.  **Feature Selection:** Use the top features identified (e.g., from Random Forest importance).
5.  **Handle Imbalance:** Apply SMOTE on the training data.
6.  **Train Model:** Use a Logistic Regression classifier.
7.  **Evaluate:** Measure performance using accuracy, precision, recall, F1-score, and a confusion matrix.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, ConfusionMatrixDisplay
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from imblearn.pipeline import Pipeline as ImbPipeline # Use imblearn's pipeline for SMOTE integration
from imblearn.over_sampling import SMOTE

# --- 1. Load Raw Data (again, to simulate starting fresh) ---
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.csv"
column_names = ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age', 'Outcome']
raw_df = pd.read_csv(url, names=column_names)

# --- Prepare Data ---
X_raw = raw_df.drop('Outcome', axis=1)
y_raw = raw_df['Outcome']

# Split into training and testing sets FIRST to prevent data leakage
X_train_raw, X_test_raw, y_train, y_test = train_test_split(X_raw, y_raw, test_size=0.3, random_state=42, stratify=y_raw)

# --- Define preprocessing steps (2 & 3: Cleaning, Imputation, Outlier Capping) ---
# We'll create a custom transformer for the 0-to-NaN and outlier capping steps
from sklearn.base import BaseEstimator, TransformerMixin

class DataCleaner(BaseEstimator, TransformerMixin):
    def __init__(self, columns_with_zero_as_missing=None):
        self.columns_with_zero_as_missing = columns_with_zero_as_missing
        self.medians = {}
        self.q1 = {}
        self.q3 = {}
        self.iqr = {}
        self.lower_bound = {}
        self.upper_bound = {}

    def fit(self, X, y=None):
        X_copy = X.copy()
        # Filter columns_with_zero_as_missing to only include those present in X_copy
        if self.columns_with_zero_as_missing:
            existing_zero_cols = [col for col in self.columns_with_zero_as_missing if col in X_copy.columns]
            X_copy[existing_zero_cols] = X_copy[existing_zero_cols].replace(0, np.nan)

        # Learn medians for imputation for existing columns
        for col in X_copy.columns:
            self.medians[col] = X_copy[col].median()

        # Learn IQR for outlier capping on all numerical columns present in X_copy
        for col in X_copy.columns:
            self.q1[col] = X_copy[col].quantile(0.25)
            self.q3[col] = X_copy[col].quantile(0.75)
            self.iqr[col] = self.q3[col] - self.q1[col]
            self.lower_bound[col] = self.q1[col] - 1.5 * self.iqr[col]
            self.upper_bound[col] = self.q3[col] + 1.5 * self.iqr[col]
        return self

    def transform(self, X):
        X_transformed = X.copy()
        # Filter columns_with_zero_as_missing to only include those present in X_transformed
        if self.columns_with_zero_as_missing:
            existing_zero_cols = [col for col in self.columns_with_zero_as_missing if col in X_transformed.columns]
            X_transformed[existing_zero_cols] = X_transformed[existing_zero_cols].replace(0, np.nan)

        # Impute missing values for existing columns
        for col in X_transformed.columns:
            if col in self.medians:
                X_transformed[col].fillna(self.medians[col], inplace=True)

        # Cap outliers for existing columns
        for col in X_transformed.columns:
            if col in self.lower_bound and col in self.upper_bound:
                X_transformed[col] = X_transformed[col].clip(lower=self.lower_bound[col], upper=self.upper_bound[col])
        return X_transformed

columns_with_zero_as_missing = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
data_cleaner = DataCleaner(columns_with_zero_as_missing=columns_with_zero_as_missing)

# --- 4. Feature Selection (using top features identified previously) ---
# From our earlier analysis, let's pick a few strong features for this mini-project
# (e.g., 'Glucose', 'BMI', 'Age', 'Insulin', 'DiabetesPedigreeFunction', 'Pregnancies')
selected_features = ['Glucose', 'BMI', 'Age', 'Insulin', 'DiabetesPedigreeFunction', 'Pregnancies']

# --- Build the Pipeline ---
# The imblearn Pipeline correctly handles fitting SMOTE only on the training data
pipeline = ImbPipeline([
    ('cleaner', data_cleaner),                     # Data Cleaning & Imputation & Outlier Capping
    ('scaler', StandardScaler()),                   # Feature Scaling
    ('smote', SMOTE(random_state=42)),              # Handle Imbalance (applied only on training data)
    ('classifier', LogisticRegression(solver='liblinear', random_state=42)) # Train Model
])

# Apply feature selection before the pipeline for both train and test sets
X_train_selected = X_train_raw[selected_features]
X_test_selected = X_test_raw[selected_features]

print("\n--- Training the full pipeline ---")
pipeline.fit(X_train_selected, y_train)

# --- 7. Evaluate the Model ---
print("\n--- Evaluating the Model Performance ---")
y_pred = pipeline.predict(X_test_selected)

print(f"Accuracy: {accuracy_score(y_test, y_pred):.4f}")
print(f"Precision: {precision_score(y_test, y_pred):.4f}")
print(f"Recall: {recall_score(y_test, y_pred):.4f}")
print(f"F1-Score: {f1_score(y_test, y_pred):.4f}")

# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(7, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['No Diabetes (0)', 'Diabetes (1)'],
            yticklabels=['No Diabetes (0)', 'Diabetes (1)'])
plt.title('Confusion Matrix of Final Pipeline Model')
plt.ylabel('Actual Label')
plt.xlabel('Predicted Label')
plt.show()

print("\n🔥 **Key takeaway for Mini Project:** A well-structured pipeline ensures that all preprocessing steps are consistently applied and, crucially, prevents data leakage by performing transformations and balancing only on the training data, then applying them to the test data. This holistic approach builds more reliable and production-ready models.")
```
